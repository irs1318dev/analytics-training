## Importing Modules
* A *module* is a file of Python code that someone else already wrote.
* Python comes with hundreds of modules, called the *standard library*.
* Before we can use a module, we have to *import* it.

```python
# The statistics module has a median function, but we haven't imported it yet.
statistics.median([1318, 254, 1678, 2910, 4911])  # CAUSES AN ERROR
```

```python
import statistics

statistics.median([1318, 254, 1678, 2910, 4911])
```

* Type the module name, a dot, then the function name.
* Put `import` statements at the top of your code, so it's easy to see which
  modules your code uses.
* In this class, always import the whole module, like `import statistics`.

### The `math` Module
Modules can hold values too, not just functions.

```python
import math

print(math.sqrt(144))
print(math.pi)
```

### The `json` Module
* JSON is a popular format for storing data as text. It looks a lot like Python
  dictionaries and lists.
* `json.loads` turns JSON text into Python dictionaries and lists.
* This is how we loaded `market_customers` in Session 3! Each line of the data
  file was JSON text for one customer.

```python
import json

robot_text = '{"team": 1318, "name": "Issaquah Robotics Society", "can_climb": true}'
robot = json.loads(robot_text)
print(type(robot_text))
print(type(robot))
robot["name"]
```

Notice that JSON uses `true` and `false`, but Python uses `True` and `False`.
`json.loads` takes care of that for us.
