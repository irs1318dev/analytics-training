### Getting Part of a List (Slicing)
* `my_list[start:stop]` gets the items from index `start` up to, but **not
  including**, index `stop`.
* Leave out `start` to begin at the beginning of the list.
* Leave out `stop` to go all the way to the end of the list.
* Slicing gives you a new list.

````python
# Items at index 1, 2, and 3, but not 4
team_numbers[1:4]
````

````python
# First 3 items
team_numbers[:3]
````

````python
# Everything from index 2 to the end
team_numbers[2:]
````

````python
# Last 2 items
team_numbers[-2:]
````

### Counting with a For Loop
* We often need to count how many items in a list meet some condition.
* Here's the pattern:
    1. Before the loop, create a counter variable and set it to 0.
    2. Inside the loop, use an `if` statement to check each item.
    3. If the item meets the condition, add 1 to the counter with `+= 1`.
* After the loop finishes, the counter holds the answer.

````python
# How many of these teams have a number >= 1000 and that is divisble by 3?
team_numbers = [1318, 254, 1678, 2910, 4911, 492, 3663]

count = 0                                   # 1. Start the counter at 0
for number in team_numbers:                 # Look at each item, one at a time
    if number % 3 == 0 and number > 1000:   # 2. Check the item
        count += 1                          # 3. Add 1 if it meets the condition
print("Teams with numbers >= 1000 and divisible by 3:", count)
````

Notice the indentation! The `if` statement is indented inside the `for` loop,
and `count += 1` is indented again inside the `if` statement. The `print`
function is not indented, so it runs once, after the loop is done.

## More on If: `else` and `elif`

* An `if` statement runs its code only when the condition is `True`.
* But what if we want to do something different when the condition is `False`?
* That's what `else` and `elif` are for.

### `else`: Do Something When the Condition Is False
* The code under `else` runs only when the `if` condition is `False`.
* `else` doesn't get a condition. It's followed by just a colon.

````python
# Robot batteries should be at least 12.5 volts before a match.
battery_voltage = 12.2

if battery_voltage >= 12.5:
    print("Battery is charged. Ready to play!")
else:
    print("Battery is low. Swap the battery!")
````

Try changing `battery_voltage` to `12.8` and running the cell again.

### `elif`: Check Another Condition
* `elif` is short for "else if."
* Use it when there are more than two possibilities.
* You can have as many `elif` sections as you want.
* Python checks the conditions from top to bottom and runs the code for the
  **first** condition that's `True`. It skips everything else.
* If none of the conditions are `True`, the `else` code runs.

````python
# Rate an alliance's match score.
alliance_score = 85

if alliance_score >= 150:
    print("Amazing match! Strategy team, take notes.")
elif alliance_score >= 100:
    print("Great match!")
elif alliance_score >= 50:
    print("Solid match. Room to improve.")
else:
    print("Time to tune the robot.")
````

**Order matters!** A score of 160 is greater than or equal to 150, 100, *and* 50,
but Python only prints "Amazing match!" because that's the first condition that's
`True`. What would happen if we checked `alliance_score >= 50` first?

### Using `elif` and `else` in a For Loop
Combine a `for` loop with `if`, `elif`, and `else` to check every item in a list.

````python
# Scouting data: number of game pieces each robot scored in a match
robots = [1318, 254, 1678, 2910, 4911]
game_pieces = [7, 12, 0, 9, 3]

for index in range(len(robots)):
    pieces = game_pieces[index]
    if pieces >= 10:
        print(robots[index], "is a scoring machine!")
    elif pieces > 0:
        print(robots[index], "scored", pieces, "game pieces.")
    else:
        print(robots[index], "didn't score. Check if their robot broke down.")
````
