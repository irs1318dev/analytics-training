## Dictionaries
In Session 2 we stored scouting data in two lists, like this:

````python
robots = [1318, 254, 1678, 2910, 4911]
game_pieces = [7, 12, 0, 9, 3]
````

How many game pieces did team 1678 score? With lists, we first have to find
1678's index in the `robots` list, then use that index in the `game_pieces` list.

````python
for index in range(len(robots)):
    if robots[index] == 1678:
        print("1678 scored", game_pieces[index], "game pieces.")
````

That works, but it's a lot of code for a simple question. A *dictionary* lets us
look up a value directly.

````python
game_pieces = {1318: 7, 254: 12, 1678: 0, 2910: 9, 4911: 3}
game_pieces[1678]
````

### Lists vs. Dictionaries

|                    | List                | Dictionary                  |
|--------------------|---------------------|-----------------------------|
| Made with          | `[]`                | `{}`                        |
| Holds              | Values              | *Key*: *value* pairs        |
| Look up items by   | Position (index)    | Key                         |
| Example            | `robots[2]`         | `game_pieces[1678]`         |

* Use a list when the order of items is what matters.
* Use a dictionary when you want to look things up by name, number, or some
  other label.

### Creating a Dictionary
* Put a colon between each key and its value.
* Separate the key-value pairs with commas.
* Keys are often strings or integers. Values can be anything, even lists or other
  dictionaries.

````python
# Splitting a long dictionary over several lines makes it easier to read.
robot = {
    "team": 1318,
    "name": "Issaquah Robotics Society",
    "drivetrain": "swerve",
    "can_climb": True,
}
robot
````

### Getting a Value
Put the key in square brackets after the dictionary.

````python
print(robot["name"])
print(robot["drivetrain"])
````

````python
# Asking for a key that isn't in the dictionary causes an error.
robot["weight"]  # CAUSES AN ERROR
````

### Checking for a Key with `in`
Use `in` to avoid the `KeyError`.

````python
print("name" in robot)
print("weight" in robot)
````

````python
if "weight" in robot:
    print("Weight:", robot["weight"])
else:
    print("Nobody weighed the robot yet.")
````

### Adding and Changing Items
* Assign a value to a new key to add it.
* Assign a value to an existing key to replace the old value. Each key can only
  appear once in a dictionary.

````python
robot["weight"] = 115         # Add a new key
robot["drivetrain"] = "tank"  # Change an existing value
robot
````

````python
# Dictionaries can start out empty, just like lists.
scores = {}
scores[254] = 12
scores[1318] = 7
print(scores, "Length:", len(scores))
````

## Looping Over Dictionaries
A `for` loop over a dictionary gives you the keys, one at a time.

````python
game_pieces = {1318: 7, 254: 12, 1678: 0, 2910: 9, 4911: 3}
for team in game_pieces:
    print(team, "scored", game_pieces[team])
````

### Looping Over Keys and Values with `items`
The `items` method gives you each key *and* its value. Put two variable names
between `for` and `in`, separated by a comma.

````python
for team, pieces in game_pieces.items():
    print(team, "scored", pieces)
````

### Looping Over Just the Values
Use the `values` method when you don't care about the keys.

````python
total = 0
for pieces in game_pieces.values():
    total += pieces
print("Total game pieces:", total)
````

### Building a Dictionary in a Loop
Start with an empty dictionary and add a key each time through the loop.

````python
# Which robots scored at least 5 game pieces?
high_scorers = {}
for team, pieces in game_pieces.items():
    if pieces >= 5:
        high_scorers[team] = pieces
high_scorers
````

### Nested Loops
* A loop can go inside another loop. The inner loop runs all the way through
  each time the outer loop runs once.
* Here, each value in `alliances` is a list. The outer loop goes over the
  alliances and the inner loop goes over the teams in each alliance.

````python
alliances = {
    "red": [1318, 254, 1678],
    "blue": [2910, 4911, 492],
}

# Build a dictionary that tells us which alliance each team is on.
team_alliance = {}
for color, teams in alliances.items():
    for team in teams:
        team_alliance[team] = color

print(team_alliance)
print("Team 4911 is on the", team_alliance[4911], "alliance.")
````

Notice that the dictionary got "flipped." `alliances` uses colors as keys, but
`team_alliance` uses team numbers as keys. Keep this trick in mind for today's
task.

## Lists of Dictionaries
* A list can hold dictionaries.
* Think of it like a table: each dictionary is a row, and the keys are the
  column names.
* In the example below, every dictionary in the list has the same keys.

````python
robots = [
    {"team": 1318, "name": "Issaquah Robotics Society", "drivetrain": "swerve"},
    {"team": 254, "name": "The Cheesy Poofs", "drivetrain": "swerve"},
    {"team": 1678, "name": "Citrus Circuits", "drivetrain": "swerve"},
    {"team": 2910, "name": "Jack in the Bot", "drivetrain": "tank"},
]
len(robots)
````

### Getting an Item
Use an index to get a dictionary from the list, then a key to get a value from
the dictionary.

````python
print("Second robot:", robots[1])
print("Name of second robot:", robots[1]["name"])
````

### Adding an Item
Use `append`, just like with any other list.

````python
robots.append({"team": 4911, "name": "CyberKnights", "drivetrain": "tank"})
len(robots)
````

### Looping Over a List of Dictionaries
Each time through the loop, the loop variable holds one dictionary.

````python
for robot in robots:
    print(robot["team"], robot["name"])
````

### Searching a List of Dictionaries
Combine a loop and an `if` statement to find the dictionaries you want.

````python
# Which teams have a tank drivetrain?
for robot in robots:
    if robot["drivetrain"] == "tank":
        print(robot["team"], "has a tank drivetrain.")
````

````python
# Save the matches in a new list instead of printing them.
swerve_robots = []
for robot in robots:
    if robot["drivetrain"] == "swerve":
        swerve_robots.append(robot)
print("Number of swerve robots:", len(swerve_robots))
````

## Writing Your Own Functions
* In Session 2 we used functions that come with Python, like `print`, `len`, and
  `int`.
* Now we'll make our own.
* Use the `def` keyword to *define* a function.

````python
def cheer():
    print("Go 1318!")
    print("Go Issaquah Robotics!")
````

* `def` is followed by the function's name, parentheses, and a colon.
* The code that belongs to the function, called the *body*, is indented.
* Running the cell above doesn't print anything! Defining a function doesn't run
  it. We have to *call* the function.

````python
cheer()
cheer()
````

### Parameters
* Put variable names inside the parentheses to let the function take arguments.
* These variable names are called *parameters*.
* When we call the function, each parameter gets the value of the matching
  argument.

````python
def cheer_for(team_number):
    print("Go", team_number)

cheer_for(1318)
cheer_for(254)
````

````python
# Separate multiple parameters with commas.
def cheer_for_alliance(team1, team2, team3):
    print("Go", team1, team2, "and", team3)

cheer_for_alliance(1318, 254, 1678)
````

````python
# Calling a function with the wrong number of arguments causes an error.
cheer_for_alliance(1318, 254)  # CAUSES AN ERROR
````

### Returning a Value
* Use `return` to send a value back to the code that called the function.
* Save the return value in a variable, just like with `int` or `len`.

````python
def match_points(game_pieces, climbed):
    points = game_pieces * 3
    if climbed:
        points += 10
    return points

score = match_points(7, True)
print("Score:", score)
print("Another score:", match_points(4, False))
````

### `print` vs. `return`
Remember: `print` displays a value, but it doesn't give it back. A function
without a `return` statement returns `None`.

````python
def match_points_printer(game_pieces, climbed):
    points = game_pieces * 3
    if climbed:
        points += 10
    print(points)

score = match_points_printer(7, True)   # Prints 31...
print("Score:", score)                  # ...but score is None!
````

### `return` Ends the Function
* Python stops running the function as soon as it reaches `return`.
* This is handy inside a loop when you've found what you're looking for.

````python
robots = [
    {"team": 1318, "name": "Issaquah Robotics Society", "drivetrain": "swerve"},
    {"team": 254, "name": "The Cheesy Poofs", "drivetrain": "swerve"},
    {"team": 2910, "name": "Jack in the Bot", "drivetrain": "tank"},
]

def find_team_name(robots, team_number):
    for robot in robots:
        if robot["team"] == team_number:
            return robot["name"]  # Found it! Stop looking.
    return None                   # Only runs if the loop never found a match.

print(find_team_name(robots, 2910))
print(find_team_name(robots, 9999))
````

### Variables Inside Functions Stay Inside
Variables created inside a function, including parameters, disappear when the
function finishes.

````python
def double(number):
    doubled = number * 2
    return doubled

print(double(21))
print(doubled)  # CAUSES AN ERROR
````

### Docstrings
* A *docstring* is a string right below the `def` line that explains what the
  function does.
* Use triple quotes. Describe the arguments and the return value.
* The `help` function displays the docstring.

````python
def count_drivetrain(robots, drivetrain):
    """Counts the robots that have a specific drivetrain.

    Args:
        robots: A list of robot dictionaries.
        drivetrain: The type of drivetrain, such as "swerve" or "tank".

    Returns:
        The number of robots with the drivetrain.
    """
    count = 0
    for robot in robots:
        if robot["drivetrain"] == drivetrain:
            count += 1
    return count

print("Swerve robots:", count_drivetrain(robots, "swerve"))
help(count_drivetrain)
````

## More String Methods
In Session 2 we used the `split` method. Strings have lots of other useful
methods.

### Changing Case with `lower` and `upper`

````python
team_name = "Issaquah Robotics Society"
print(team_name.lower())
print(team_name.upper())
````

````python
# Python is case sensitive, so "A" and "a" are different strings.
print("A" == "a")
print("A".lower() == "a")
````

### String Methods Don't Change the Original String
* String methods give you a *new* string. The original string stays the same.
* To keep the result, save it in a variable.

````python
team_name.lower()
print(team_name)  # Still has capital letters!

lower_name = team_name.lower()
print(lower_name)
````

### Replacing Text with `replace`
`replace` swaps every copy of the first argument with the second argument.

````python
phone = "425-555-0123"
print(phone.replace("-", "."))
print(phone.replace("-", ""))  # Replace with an empty string to remove it.
````

### Checking Characters with `isalpha` and `isdigit`
* `isalpha` returns `True` if every character is a letter.
* `isdigit` returns `True` if every character is a digit (0 through 9).

````python
print("a", "a".isalpha())
print("-", "-".isalpha())
print("' '", " ".isalpha())
print("7", "7".isdigit())
print("1318", "1318".isdigit())
````

### Building a String in a Loop
* Start with an empty string, then add to it with `+=`.
* This works just like adding to a number, except it adds characters to the end
  of the string.

````python
# Keep only the letters from a name.
name = "O'Brien-Smith Jr."
letters = ""
for character in name:
    if character.isalpha():
        letters += character
letters
````

### Converting Numbers to Strings with `str`
* `str` isn't a method, it's a function, like `int` and `float`.
* You can't use `+` to join a string and a number. Convert the number first.

````python
team_number = 1318
"Team " + team_number  # CAUSES AN ERROR
````

````python
"Team " + str(team_number)
````

### Checking the Start and End with `startswith` and `endswith`

````python
name = "Joseph Miller Jr."
print(name.startswith("Jo"))
print(name.endswith("Jr."))
````

### Searching a String with `in`
`in` works with strings too. It checks if one string is inside another.

````python
print("Robotics" in team_name)
print("robotics" in team_name)  # Remember, case matters!
````
